```python
data = input("Masukkan nama, NIM, nilai tugas, nilai UTS, nilai UAS: ")
nama, nim, nilai_tugas, nilai_uts, nilai_uas = data.split(",")

nama = nama.strip()
nim = nim.strip()
nilai_tugas = float(nilai_tugas.strip())
nilai_uts = float(nilai_uts.strip())
nilai_uas = float(nilai_uas.strip())

bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45
nilai_akhir = (nilai_tugas * bobot_tugas
               + nilai_uts * bobot_uts
               + nilai_uas * bobot_uas)
bobot_total = bobot_tugas + bobot_uts + bobot_uas

lebar = max(58, len(nama) + 9, len(nim) + 9)
lebar_komponen = lebar - 28
garis = "+" + "-" * lebar + "+"

print()
print(garis)
print(f"|{'RINGKASAN NILAI AKADEMIK':^{lebar}}|")
print(garis)
print(f"|{(' Nama : ' + nama):<{lebar}}|")
print(f"|{(' NIM  : ' + nim):<{lebar}}|")
print(garis)
print(f"| {'Komponen':<{lebar_komponen}} {'Nilai':>12} {'Bobot':>12} |")
print(garis)
print(f"| {'Tugas':<{lebar_komponen}} {nilai_tugas:>12.1f} {bobot_tugas:>12.0%} |")
print(f"| {'UTS':<{lebar_komponen}} {nilai_uts:>12.1f} {bobot_uts:>12.0%} |")
print(f"| {'UAS':<{lebar_komponen}} {nilai_uas:>12.1f} {bobot_uas:>12.0%} |")
print(garis)
print(f"| {'Nilai Akhir':<{lebar_komponen}} {nilai_akhir:>12.1f} {bobot_total:>12.0%} |")
print(garis)
```
